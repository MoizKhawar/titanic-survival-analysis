```python
# importing necessary libraries
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#import titanic dataset
titanic = sns.load_dataset('titanic')
# display the first 5 rows of the dataset
titanic.head()
```

```python
#calculate survival rate base on gender
survival_rate = titanic['survived'].groupby(titanic['sex']).mean()
print(survival_rate)
```

```python
# set figure size
plt.figure(figsize=(4, 3))

# create bar plot
sns.barplot(x=survival_rate.index, y=survival_rate.values)
plt.title('Survival Rate')
plt.ylabel('Survival Rate')
plt.xlabel('Gender')

plt.show()
```

```python
# create pie chart
plt.figure(figsize=(4, 3))
plt.pie(survival_rate, labels=survival_rate.index, autopct='%1.1f%%', startangle=90)    
plt.title('Survival Rate by Gender')
plt.ylabel('Survival Rate')
plt.xlabel('Gender')
plt.show()
```

```python
#  Observation
# 1. The survival rate for females was significantly higher than for males, with around 74% of women surviving compared to only 19% of men.
# 2. This suggests that gender played a crucial role in survival during the Titanic disaster.
# 3. The data may reflect societal norms of the time
# 4. The higher survival rate for women may indicate the prioritization of women and children in rescue efforts.
# 5. The data may also reflect the fact that many men were left behind during the evacuation.
# 6. Overall, the data suggests that societal factors significantly influenced survival outcomes during the Titanic disaster.
# 7. The data may also reflect the fact that many men were unable to access lifeboats due to overcrowding.
# 8. The data may also reflect the fact that many men were reluctant to leave their posts.
# 9. This highlights the complex interplay between gender roles and survival in crisis situations.
```
